# Final analysis and conclusion

In this project, we analyzed insurance customer reviews using different natural language processing approaches.

We addressed two main supervised learning tasks:
- Rating prediction (1 to 5 stars)
- Sentiment analysis (positive, neutral, negative)

We compared multiple models, from classical machine learning methods to modern embedding based and transformer-based approaches.

### Results Overview
The following table summarizes the main results obtained for rating prediction:

```python
import pandas as pd

rating_results = {
    "TF-IDF + Logistic Regression": 0.51,
    "Embedding Layer Model": 0.526,
    "Pre-trained Embeddings": 0.464,
    "SBERT + Logistic Regression": 0.524,
    "BERT (subset)": 0.54,
    "LLM": 0.20
}

df_rating = pd.DataFrame(
    list(rating_results.items()),
    columns=["Model", "Accuracy"]
).sort_values(by="Accuracy", ascending=True)

df_rating
```

## Model Comparison for Rating Prediction

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.barh(df_rating["Model"], df_rating["Accuracy"])
plt.xlabel("Accuracy")
plt.title("Rating Prediction Model Comparison")
plt.show()
```

## Interpretation of Results

Rating prediction appears to be a difficult task.
All classical TF-IDF-based models achieve similar performance which indicates that most of the useful signal is already captured by word frequencies. However, these models struggle to distinguish between close ratings such as 3 and 4 stars.
Neural and embedding based approaches provide only moderate improvements. The embedding layer model slightly improves performance while SBERT shows that sentence level representations are more informative than simple bag of words features.

The best result is obtained with the pre trained BERT model, confirming that contextual embeddings are the most effective for capturing nuanced meaning in customer reviews.
In contrast, the model based on averaged pre trained embeddings performs worse, as this approach loses important contextual information. The LLM based approach also performs poorly, suggesting that prompting alone is not sufficient for precise multi class prediction.

## Rating Prediction vs Sentiment Analysis

Sentiment analysis achieves significantly higher accuracy than rating prediction.
This is because sentiment classification is a simpler task where reviews are grouped into broader categories. In contrast, rating prediction requires distinguishing between fine grained classes, which is much more challenging.
This result highlights that while predicting exact ratings is difficult, identifying general customer satisfaction is more reliable and easier to model.

## Key Insights

- Customer reviews contain nuanced opinions that are difficult to map to exact ratings
- Classical models already provide strong baselines
- Contextual models (SBERT, BERT) improve performance but do not fully solve the problem
- Averaging embeddings leads to loss of important information
- LLM-based approaches require more advanced prompting or fine tuning

## Limitations

- Some experiments were performed on subsets due to computational constraints
- BERT was not fully fine tuned on the entire dataset
- The LLM evaluation was limited to simple prompting
- The dataset may contain noisy or ambiguous reviews

## Future work we can do to improve our models 

- Fine-tune transformer models on the full dataset
- Improve LLM prompting strategies
- Explore regression instead of classification for ratings
- Use more advanced embedding aggregation methods
- Integrate additional features such as metadata

## Final Conclusion

This project shows that analyzing customer reviews is both valuable and challenging.
While classical machine learning models provide strong baselines, more advanced models based on contextual embeddings offer better performance.
However, predicting exact ratings remains difficult due to the subtle differences between classes. In contrast, sentiment analysis proves to be more robust and reliable.
Overall, transformer based models represent the most promising direction, but further improvements require fine tuning and more advanced modeling strategies.
